```python
def process(res: SwimDataDict):
    init_mask = filt_init_mask(res['behav_pos_y'], res['behav_time'])
    for k in res.keys():
        res[k] = res[k][init_mask]
        
    # Smooth both channel:
    for k in ['fltCh0', 'fltCh1']:
        res[k] = np.convolve(res[k], np.ones(600)/600, mode='same')
    
    # Compute speed:
    x = res['behav_pos_y'] + 50 * res['n_trials']
    t = res['behav_time']
    dx = np.diff(x)
    dt = np.diff(t)
    
    idx = np.append([0], np.where(dx/dt >= 0)[0])  # Threshold for movement, can be adjusted
    for k in res.keys():
        res[k] = res[k][idx]    
    
    dx = dx[idx[1:]]  # Adjust dx to match the length of idx
    dt = dt[idx[1:]]  # Adjust dt to match the length of idx
    
    speed = np.convolve(dx/dt, np.ones(600)/600, mode='same')
    speed = np.concatenate([[0], speed])  # Pad the first element to match the length of other arrays
    res['smooth_speed'] = speed
    

    
    # Downsample fro 6kHz to 30Hz:
    for k in res.keys():
        res[k] = res[k][::200]    
    
    for k in res.keys():
        res[k] = res[k][10:-10]
        
    return res
```

# Spatial Map Clustering

```python
# Compute CI per bin
def compute_CI(x, y, n_bin=100):
    top_boundary = np.zeros((y.shape[0], n_bin))
    low_boundary = np.zeros((y.shape[0], n_bin))
    means = np.zeros((y.shape[0], n_bin))
    for i in tqdm(range(n_bin)):
        bin_values = y[:, x == i]
        if len(bin_values) > 0:
            mean = np.nanmean(bin_values, axis=1)
            means[:, i] = mean
            sem = np.nanstd(bin_values, axis=1) / np.sqrt(bin_values.shape[1])
            top_boundary[:, i] = mean + sem
            low_boundary[:, i] = mean - sem
        else:
            top_boundary[:, i] = np.nan
            low_boundary[:, i] = np.nan

    x = np.arange(n_bin)/n_bin * 2 * np.pi
    y = means
    return x, y, low_boundary, top_boundary

save_dir = os.path.join(path, "SpatialMapClustering")
mkdir(save_dir)
x = trace['spike_nodes'][200:]
y = trace['RawTraces'][:, 200:]

_, y_map, l, t = compute_CI(x, y, n_bin=trace['rate_map_all'].shape[1])
sem_map = t - l
    
total_maps = np.concatenate([y_map, sem_map], axis=1)
total_maps = total_maps / np.max(y_map, axis=1, keepdims=True) # Normalize by max response for better visualization
```

```python
from sklearn.cluster import DBSCAN, KMeans
from sklearn.decomposition import PCA   

qualified_cells = np.where(trace['p_values'] < 0.005)[0]

kmeans = KMeans(n_clusters=20, random_state=0)
kmeans.fit(total_maps[qualified_cells])
clusters = kmeans.labels_.astype(np.int32)
clusters -= np.min(clusters)  # Shift cluster labels to start from 0
print(np.unique(clusters), clusters.shape)
color = sns.color_palette("rainbow", np.max(clusters)+1)
print(f"Clustering down: \n  {np.histogram(clusters, bins=np.arange(-0.5, np.max(clusters)+1.5, 1))[0]}")

# Compute mean tuning curve for each cluster
centroid_map = np.zeros((np.max(clusters)+1, y_map.shape[1]))
for i in tqdm(range(np.max(clusters)+1)):
    centroid_map[i] = np.mean(y_map[qualified_cells][clusters == i], axis=0)

pca = PCA(n_components=2)
pca_result = pca.fit_transform(total_maps[qualified_cells])

plt.figure(figsize=(6, 6))
ax = Clear_Axes(plt.axes(), close_spines=['top', 'right'], ifxticks=True, ifyticks=True)
ax.scatter(
    pca_result[:, 0], 
    pca_result[:, 1], 
    c=clusters,
    cmap='rainbow', 
    alpha=0.7, 
    s=2,
    edgecolors=None
)
plt.show()

def visualize_cluster(cluster_id, rois, ):
    plt.figure(figsize=(4, 3))
    ax = Clear_Axes(plt.axes(), close_spines=['top', 'right'], ifxticks=True, ifyticks=True)
    sns.lineplot(x=np.arange(y_map.shape[1]), y=centroid_map[cluster_id], ax=ax, color=color[cluster_id])
    ylims = ax.get_ylim()
    for (start, end) in landmark_pos:
        ax.fill_betweenx(ylims, start*100, end*100, color=ChannelColors[0], alpha=0.1, edgecolor=None)      
    ax.set_xlabel('Position (%)')  
    ax.set_ylabel('Normalized dF/F')
    ax.set_xticks(np.linspace(0, 100, 6))
    ax.set_title(f"Cluster {cluster_id} (n={np.sum(clusters==cluster_id)})")
    plt.tight_layout()
    plt.savefig(os.path.join(save_dir, f"Cluster {cluster_id}.svg"), dpi=300)
    plt.show()
```

```python
from scipy.ndimage import gaussian_filter
rois = cp.deepcopy(trace['roi_coord'])

def calc_density(
    rois: np.ndarray,
    stype: str = 'xy',
    nx = 100,
    ny = 100
):
    if stype == 'xy':
        x, y = rois[:, 0], rois[:, 1]
    elif stype == 'xz':
        x, y = rois[:, 0], rois[:, 2]
    elif stype == 'yz':
        x, y = rois[:, 1], rois[:, 2]
    else:
        raise ValueError("stype must be one of 'xy', 'xz', 'yz'")
    
    heatmap, xedges, yedges = np.histogram2d(x, y, bins=[nx, ny], range=[[0-0.5, nx-0.5], [0-0.5, ny-0.5]], density=True)
    # smooth the heatmap
    
    heatmap = gaussian_filter(heatmap, sigma=5)
    return heatmap

def visualize_cluster(
    rois: np.ndarray, 
    meanImg: np.ndarray,
    color: str = ChannelColors[0]
):
        
    fig = plt.figure(figsize=(9, 9))
    ax = fig.add_subplot(111, projection='3d')
    # 3d scatter plot of ROIs colored by cluster
    rois[:, 0] = rois[:, 0] % trace['lenX'] + np.random.rand(rois.shape[0]) - 0.5
    rois[:, 1] = rois[:, 1] % trace['lenY'] + np.random.rand(rois.shape[0]) - 0.5
    scatter = ax.scatter(
        rois[:, 0], 
        rois[:, 1],  
        rois[:, 2],
        c=color,
        alpha=1, 
        s=3,
        edgecolors=None
    )
    
    # project 2d planes onto the bottom
    #meanImg = meanImg[::4, :, :] + meanImg[1::4, :, :] + meanImg[2::4, :, :] + meanImg[3::4, :, :]
    #meanImg = meanImg[:, ::4, :] + meanImg[:, 1::4, :] + meanImg[:, 2::4, :] + meanImg[:, 3::4, :]
    nx, ny, nz = meanImg.shape
    print(nx, ny, nz)
    
    z_plane = nz+5
    x_plane = nx+5
    y_plane = ny+5
    
    X_xy, Y_xy = np.meshgrid(np.arange(nx), np.arange(ny), indexing='ij')
    img_xy_n = np.median(meanImg, axis=2)
    Z_xy = np.zeros_like(X_xy)+z_plane
    ax.plot_surface(
        Y_xy, X_xy, Z_xy,
        facecolors=sns.color_palette("gray", as_cmap=True)(img_xy_n / np.max(img_xy_n)),
    )   
    density_xy = calc_density(rois, stype='xy', nx=ny, ny=nx)
    density_xy[density_xy < 1e-8] = np.nan
    ax.contourf(
        Y_xy, X_xy, density_xy.T, 
        zdir='z', offset=z_plane-1, cmap='Greens'
    )
    ax.set_xlim(0, trace['lenX'])
    ax.set_ylim(0, trace['lenY'])
    ax.set_zlim(0, z_plane)
    ax.invert_zaxis()
    plt.show()

meanImg = []
px, py = int(trace['meanImg'].shape[1] / trace['lenX']), int(trace['meanImg'].shape[0] / trace['lenY'])
for i in range(trace['nplanes']):
    nr, nc = i // px, i % px
    meanImg.append(trace['meanImg'][nr*trace['lenY']:(nr+1)*trace['lenY'], nc*trace['lenX']:(nc+1)*trace['lenX']])
meanImg = np.stack(meanImg, axis=2)


visualize_cluster(rois[qualified_cells][clusters == 1], meanImg)
```
